# Lab | Chains in LangChain — Complete Solution

This notebook solves the uploaded lab using **LangChain Expression Language (LCEL)**, the modern replacement for legacy `LLMChain`, `SimpleSequentialChain`, `SequentialChain`, and `MultiPromptChain` APIs. It preserves the four lab sections and includes multiple product/review examples.

**Requirements:** Python 3.10+, OpenAI API key, internet access for package installation and model calls. Running model cells may incur API charges. Outputs are not pre-generated.

## 0. Setup and dataset

```python
import os
from pathlib import Path
import pandas as pd
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser, JsonOutputParser
from langchain_core.runnables import RunnableLambda, RunnablePassthrough

load_dotenv()
if not os.getenv("OPENAI_API_KEY"):
    raise EnvironmentError("Set OPENAI_API_KEY in your environment or a local .env file before running model calls.")

# Change this if your account uses a different available model.
MODEL = os.getenv("OPENAI_MODEL", "gpt-4o-mini")
llm = ChatOpenAI(model=MODEL, temperature=0.3)
parser = StrOutputParser()
print("Model:", MODEL)
```

```python
# Works with the uploaded files in the same folder or the original ./data/ folder.
paths = [Path("Data.csv"), Path("data/Data.csv"), Path("/mnt/data/Data.csv")]
csv_path = next((p for p in paths if p.exists()), None)
if csv_path is None:
    raise FileNotFoundError("Place Data.csv next to this notebook (or in ./data/).")
df = pd.read_csv(csv_path)
assert {"Product", "Review"}.issubset(df.columns)
df = df.dropna(subset=["Product", "Review"]).reset_index(drop=True)
display(df.head())
print(f"Loaded {len(df)} reviews")
```

## 1. LLMChain (modern LCEL equivalent)

**Original exercise:** choose a temperature, build a prompt that accepts a product type, and generate a description.

**Choice:** `temperature=0.3` favors consistent, grounded descriptions. We instruct the model not to invent specific specifications. LCEL uses `prompt | llm | parser` rather than the deprecated `LLMChain`.

```python
description_prompt = ChatPromptTemplate.from_template("""You are a professional e-commerce copywriter.
Write a concise, engaging description (60-90 words) for the product category: {product}.
Explain common uses and possible benefits, but do not invent brand-specific features,
materials, certifications, prices, or performance claims. Use clear English.""")
description_chain = description_prompt | llm | parser
product = str(df.loc[0, "Product"])
print("PRODUCT:", product)
print(description_chain.invoke({"product": product}))
```

## 2. SimpleSequentialChain (two consecutive steps)

Step 1 produces a product description; step 2 uses that description to generate a short marketing pitch. The output of the first chain is automatically passed to the second.

```python
creative_llm = ChatOpenAI(model=MODEL, temperature=0.8)
first_prompt = ChatPromptTemplate.from_template("""Describe the product category '{product}' in 3 useful sentences.
Avoid unverified technical or brand-specific claims.""")
second_prompt = ChatPromptTemplate.from_template("""Turn the following product description into a catchy,
trustworthy two-sentence marketing pitch with a short headline. Do not invent claims.
Description: {description}""")
chain_one = first_prompt | creative_llm | parser
chain_two = second_prompt | creative_llm | parser
simple_sequential_chain = chain_one | RunnableLambda(lambda description: {"description": description}) | chain_two
print(simple_sequential_chain.invoke({"product": product}))
```

### Repeat for two additional products

```python
products = df["Product"].drop_duplicates().head(3).tolist()
for item in products:
    print("\n" + "=" * 60)
    print("PRODUCT:", item)
    print(simple_sequential_chain.invoke({"product": item}))
```

## 3. SequentialChain (four connected steps)

The original lab requests: **translate a review**, **summarize it**, **identify its language**, and **generate a follow-up message using earlier outputs**.

Here the review is translated to English, summarized, the original language is detected, and a customer-service reply is created from the translation, summary, and language. `RunnablePassthrough.assign` retains all intermediate values, analogous to a multi-output SequentialChain.

```python
review_llm = ChatOpenAI(model=MODEL, temperature=0.2)
translation_prompt = ChatPromptTemplate.from_template("""Translate this customer review into English.
If it is already English, preserve its meaning. Output only the English review.
Review: {Review}""")
summary_prompt = ChatPromptTemplate.from_template("""Summarize this customer review in 1-2 sentences.
Mention key praise and complaints, without inventing details.
English review: {English_Review}""")
language_prompt = ChatPromptTemplate.from_template("""Identify the original language of this review.
Reply with only the language name (e.g. English, French, German).
Review: {Review}""")
followup_prompt = ChatPromptTemplate.from_template("""You are a helpful customer service representative.
Write a polite, empathetic follow-up response in English (3-5 sentences).
Acknowledge the customer's experience; do not promise refunds or actions not authorized.
Original language: {language}
English review: {English_Review}
Summary: {summary}""")
translation_chain = translation_prompt | review_llm | parser
summary_chain = summary_prompt | review_llm | parser
language_chain = language_prompt | review_llm | parser
followup_chain = followup_prompt | review_llm | parser

overall_chain = (
    RunnablePassthrough.assign(English_Review=translation_chain)
    | RunnablePassthrough.assign(summary=summary_chain)
    | RunnablePassthrough.assign(language=language_chain)
    | RunnablePassthrough.assign(followup_message=followup_chain)
)
review = str(df.loc[5, "Review"])
result = overall_chain.invoke({"Review": review})
for key in ["English_Review", "summary", "language", "followup_message"]:
    print(f"\n{key}:\n{result[key]}")
```

### Repeat for two additional reviews

```python
for idx in [0, 1]:
    item = df.loc[idx]
    print("\n" + "=" * 70)
    print("PRODUCT:", item["Product"])
    output = overall_chain.invoke({"Review": str(item["Review"])})
    for key in ["English_Review", "summary", "language", "followup_message"]:
        print(f"\n{key}: {output[key]}")
```

## 4. Router Chain (modern replacement for MultiPromptChain)

The original notebook defines five specialist prompts. We reuse **those original templates** and select the best one using structured model output. The router can choose `DEFAULT` for unrelated questions. Routing uses a temperature of zero for consistency.

```python
physics_template = """You are a very smart physics professor. \
You are great at answering questions about physics in a concise\
and easy to understand manner. \
When you don't know the answer to a question you admit\
that you don't know.

Here is a question:
{input}"""


math_template = """You are a very good mathematician. \
You are great at answering math questions. \
You are so good because you are able to break down \
hard problems into their component parts, 
answer the component parts, and then put them together\
to answer the broader question.

Here is a question:
{input}"""

history_template = """You are a very good historian. \
You have an excellent knowledge of and understanding of people,\
events and contexts from a range of historical periods. \
You have the ability to think, reflect, debate, discuss and \
evaluate the past. You have a respect for historical evidence\
and the ability to make use of it to support your explanations \
and judgements.

Here is a question:
{input}"""


computerscience_template = """ You are a successful computer scientist.\
You have a passion for creativity, collaboration,\
forward-thinking, confidence, strong problem-solving capabilities,\
understanding of theories and algorithms, and excellent communication \
skills. You are great at answering coding questions. \
You are so good because you know how to solve a problem by \
describing the solution in imperative steps \
that a machine can easily interpret and you know how to \
choose a solution that has a good balance between \
time complexity and space complexity. 

Here is a question:
{input}"""

biology_template = """You are an excellent biologist. \
You have a deep understanding of living organisms, \
from the molecular and cellular level to entire ecosystems. \
You are skilled at observing patterns in nature, analyzing biological data, \
and explaining complex processes like evolution, genetics, physiology, and ecology. \
You can clearly communicate how life functions and adapts, \
and you make connections between different biological concepts \
to answer challenging questions.

Here is a question:
{input}"""
```

```python
prompt_infos = [
    {"name": "physics", "description": "Physics and physical phenomena", "prompt_template": physics_template},
    {"name": "math", "description": "Mathematics, algebra and calculations", "prompt_template": math_template},
    {"name": "history", "description": "Historical people, events and periods", "prompt_template": history_template},
    {"name": "computer science", "description": "Programming, algorithms and computer science", "prompt_template": computerscience_template},
    {"name": "biology", "description": "Living organisms, cells, genetics and ecosystems", "prompt_template": biology_template},
]
router_llm = ChatOpenAI(model=MODEL, temperature=0)
destination_chains = {
    p["name"]: ChatPromptTemplate.from_template(p["prompt_template"]) | router_llm | parser
    for p in prompt_infos
}
default_chain = ChatPromptTemplate.from_template("Answer this question clearly and concisely: {input}") | router_llm | parser
names = list(destination_chains) + ["DEFAULT"]
destinations = "\n".join(f"- {p['name']}: {p['description']}" for p in prompt_infos)
route_prompt = ChatPromptTemplate.from_template("""Choose the most appropriate specialist for the question.
Available destinations:
{destinations}

Question: {input}
Return ONLY a JSON object with the fields "destination" and "next_inputs".
"destination" must be one of: {names}.
Use DEFAULT if none fits. "next_inputs" should preserve the question's meaning.""")
route_model = route_prompt | router_llm | JsonOutputParser()

def route_question(question: str) -> dict:
    decision = route_model.invoke({"input": question, "destinations": destinations, "names": ", ".join(names)})
    destination = str(decision.get("destination", "DEFAULT")).strip()
    if destination not in destination_chains:
        destination = "DEFAULT"
    next_input = decision.get("next_inputs") or question
    chosen_chain = destination_chains.get(destination, default_chain)
    answer = chosen_chain.invoke({"input": str(next_input)})
    return {"question": question, "destination": destination, "answer": answer}

questions = [
    "What is black body radiation?",
    "What is 2 + 2?",
    "Why does every cell in our body contain DNA?",
]
for q in questions:
    result = route_question(q)
    print(f"\nQUESTION: {q}\nROUTE: {result['destination']}\nANSWER: {result['answer']}")
```

### Additional creative routing tests

```python
extra_questions = [
    "What caused the fall of the Roman Empire?",
    "How does binary search work, and what is its time complexity?",
    "How does photosynthesis convert sunlight into chemical energy?",
    "Why does a rainbow form after rain?",
    "What are three good habits for organizing a study schedule?",  # DEFAULT
]
for question in extra_questions:
    result = route_question(question)
    print(f"\n[{result['destination']}] {question}\n{result['answer']}")
```

## Conclusion

| Original lab concept | Modern LangChain solution |
|---|---|
| `LLMChain` | `prompt | llm | StrOutputParser()` |
| `SimpleSequentialChain` | LCEL pipeline using `|` and `RunnableLambda` |
| `SequentialChain` | `RunnablePassthrough.assign(...)` for named intermediate outputs |
| `MultiPromptChain` / `LLMRouterChain` | Structured JSON router + specialist LCEL chains |

**Why use LCEL?** It avoids removed legacy imports, makes input/output flow explicit, and supports modular composition. The model responses are nondeterministic and will vary. To execute, set your `OPENAI_API_KEY` and run the cells in order. The notebook does not include fabricated API results.
